## Project Introduction

This project analyses 14 days of sales data from Wanjiku's fresh-produce stall at Gikomba Market. The analysis will help identify sales trends, compare weekday and weekend performance, monitor stock levels, and calculate product profit margins.

The goal is to use Python to understand the business performance and provide practical recommendations to help Wanjiku make better business decisions.

```python
# Daily sales for 14 days (KES)
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]

# Days of the week (repeated for two weeks)
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# Stock information
# qty = quantity in stock (kg)
# reorder = reorder level (kg)
# cost and price are in KES per kg

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

 Task 1: Basic Sales Analysis

In this task, I will use Python's built-in functions to calculate total sales, average sales, highest sales, and lowest sales over the 14-day period. I will also identify the best and worst sales days.

```python
# Calculate total, average, highest and lowest sales

total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

# Find the days with the highest and lowest sales
best_day_index = daily_sales.index(highest_sales)
worst_day_index = daily_sales.index(lowest_sales)

best_day = days[best_day_index]
worst_day = days[worst_day_index]

# Display the results
print("BASIC SALES ANALYSIS")
print("--------------------")
print(f"Total sales: KES {total_sales:,.2f}")
print(f"Average daily sales: KES {average_sales:,.2f}")
print(f"Highest sales: KES {highest_sales:,.2f}")
print(f"Lowest sales: KES {lowest_sales:,.2f}")
print(f"Best sales day: {best_day}")
print(f"Worst sales day: {worst_day}")
```

Task 1 (continued): Weekly Sales Comparison

I will divide the 14 days into two groups of seven days. I will calculate the total sales for each week and identify which week performed better.

```python
# Separate sales into two weeks
week1_sales = daily_sales[:7]
week2_sales = daily_sales[7:]

# Calculate weekly totals
week1_total = sum(week1_sales)
week2_total = sum(week2_sales)

# Compare the two weeks
print("WEEKLY SALES COMPARISON")
print("-----------------------")
print(f"Week 1 sales: KES {week1_total:,.2f}")
print(f"Week 2 sales: KES {week2_total:,.2f}")

if week2_total > week1_total:
    print("Better performing week: Week 2")
else:
    print("Better performing week: Week 1")
```

Weekly Sales Growth

I will calculate the percentage change in sales from Week 1 to Week 2 to measure whether Wanjiku's business improved or declined.

```python
# Calculate weekly sales growth
growth = ((week2_total - week1_total) / week1_total) * 100

print("WEEKLY SALES GROWTH")
print("-------------------")
print(f"Sales growth: {growth:.2f}%")
```

 Task 2: Weekly Sales Function

I will create a reusable function called `week_totals()` that uses list slicing to separate the first seven days from the next seven days and returns the total sales for each week as a tuple.

```python
def week_totals(sales):
    """
    Calculate total sales for Week 1 and Week 2.

    Args:
        sales: A list containing 14 daily sales amounts.

    Returns:
        A tuple containing Week 1 and Week 2 sales totals.
    """
    week1 = sum(sales[:7])
    week2 = sum(sales[7:14])

    return (week1, week2)


# Call the function
weekly_results = week_totals(daily_sales)

print("RESULTS FROM week_totals()")
print("--------------------------")
print(f"Week 1: KES {weekly_results[0]:,}")
print(f"Week 2: KES {weekly_results[1]:,}")
```

Task 3: Daily Sales Classification

I will create a function called `classify_day()` to classify each day's sales as Strong, Normal, or Weak based on the average daily sales. I will then count how many days fall into each category using a dictionary.

```python
def classify_day(amount, average):
    """
    Classify daily sales based on the average sales amount.

    Args:
        amount: Sales amount for one day.
        average: Average daily sales.

    Returns:
        Strong, Normal, or Weak.
    """
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"


# Create a dictionary to count each category
sales_categories = {}

# Classify each day's sales
for amount in daily_sales:
    category = classify_day(amount, average_sales)

    sales_categories[category] = sales_categories.get(category, 0) + 1

# Display the results
print("DAILY SALES CLASSIFICATION")
print("--------------------------")

for category, count in sales_categories.items():
    print(f"{category}: {count} days")
```

Task 4: Weekday vs Weekend Sales Analysis

I will separate sales into weekdays (Monday to Friday) and weekends (Saturday and Sunday). I will calculate the average sales for each group and compare their performance using a percentage difference.

```python
# Separate weekday and weekend sales
weekday_sales = []
weekend_sales = []

for i in range(len(daily_sales)):
    if days[i] in ["Mon", "Tue", "Wed", "Thu", "Fri"]:
        weekday_sales.append(daily_sales[i])
    else:
        weekend_sales.append(daily_sales[i])

# Calculate average sales
weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

# Calculate percentage difference
percentage_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100

# Display results
print("WEEKDAY VS WEEKEND ANALYSIS")
print("---------------------------")
print(f"Weekday average: KES {weekday_average:,.2f}")
print(f"Weekend average: KES {weekend_average:,.2f}")
print(f"Percentage difference: {percentage_difference:.2f}%")

if weekend_average > weekday_average:
    print("Better performance: Weekends")
else:
    print("Better performance: Weekdays")
```

Task 5: Stock Management and Gross Margins

I will create a function to identify products whose current stock quantity is at or below their reorder level. I will also calculate the gross margin percentage for each product using its cost and selling price.

```python
def products_to_reorder(stock):
    """
    Identify products whose quantity is at or below reorder level.

    Args:
        stock: Dictionary containing product stock information.

    Returns:
        A list of product names that need restocking.
    """
    reorder_list = []

    for product, details in stock.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)

    return reorder_list


# Find products that need restocking
reorder_products = products_to_reorder(stock)

print("PRODUCTS NEEDING RESTOCKING")
print("---------------------------")

for product in reorder_products:
    print(product)
```

Gross Profit Margin Analysis

I will calculate the gross margin percentage for each product using its selling price and cost price. This will help identify which products provide the highest gross margin.

```python
print("PRODUCT GROSS MARGINS")
print("---------------------")

for product, details in stock.items():
    cost = details["cost"]
    price = details["price"]

    margin = ((price - cost) / price) * 100

    print(f"{product}: {margin:.2f}%")
```

```python
# STALL PERFORMANCE REPORT

print("=" * 45)
print("       W A N J I K U ' S  S T A L L")
print("          PERFORMANCE REPORT")
print("=" * 45)

print("\n1. SALES SUMMARY")
print("-" * 45)
print(f"Total sales:       KES {total_sales:,.2f}")
print(f"Average daily:     KES {average_sales:,.2f}")
print(f"Highest sales:     KES {highest_sales:,.2f}")
print(f"Lowest sales:      KES {lowest_sales:,.2f}")
print(f"Best sales day:    {best_day}")
print(f"Worst sales day:   {worst_day}")

print("\n2. WEEKLY PERFORMANCE")
print("-" * 45)
print(f"Week 1:            KES {week1_total:,.2f}")
print(f"Week 2:            KES {week2_total:,.2f}")
print(f"Weekly growth:     {growth:.2f}%")

print("\n3. DAILY SALES CLASSIFICATION")
print("-" * 45)
for category, count in sales_categories.items():
    print(f"{category:<12} {count} days")

print("\n4. WEEKDAY VS WEEKEND")
print("-" * 45)
print(f"Weekday average:   KES {weekday_average:,.2f}")
print(f"Weekend average:   KES {weekend_average:,.2f}")
print(f"Difference:        {percentage_difference:.2f}%")

print("\n5. STOCK ALERTS")
print("-" * 45)
for product in reorder_products:
    print(f"REORDER: {product}")

print("\n6. GROSS PROFIT MARGINS")
print("-" * 45)
for product, details in stock.items():
    margin = ((details["price"] - details["cost"])
              / details["price"]) * 100
    print(f"{product:<12} {margin:.2f}%")

print("\n" + "=" * 45)
print("             END OF REPORT")
print("=" * 45)
```
